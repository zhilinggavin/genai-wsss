### This code in temparary, used to categrise the images into different folders based on patient ID


```python
from PIL import Image
import os
import glob
import shutil
from tqdm import tqdm
img_dir = '/media/NAS06/gavinyue/genai-wsss/experiments/full_supervised_unet/results/AIPFR/imgs'
save_dir = '/media/NAS06/gavinyue/genai-wsss/experiments/full_supervised_unet/results/AIPFR/pred_mask'
files = glob.glob(img_dir + '/*.png')
print(len(files))
```

```python
for file in tqdm(files):
    # slice_name: case02_00019_slice100_mask
    # get case name, e.g. '02_00019'
    slice_name = file.split('/')[-1].split('.')[0]
    # print(slice_name)
    case = (slice_name.split('_slice')[0]).replace('case', '')
    # print(case)
    
    # make dir of case
    case_dir = os.path.join(save_dir, case)
    os.makedirs(case_dir, exist_ok=True)
    
    # move file to case dir
    shutil.move(file, case_dir)

print('all slice images moved to case dirs')
```
