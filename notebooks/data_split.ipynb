# Split dataset into Train, Val, Test

## OSIC dataset split. 
1. Split patient cases based on each doctor. For example, leave 20% of patients from each doctor, resulting in 20% of the cases being used for testing.
2. Train 64% : Val 16% : Test 20%

### Split Test 3d cases

```python
# read doctor metadata
import pandas as pd
df = pd.read_csv('../data/OSIC/doctor_category.csv')
doctors = df['doctor'].unique()
case_id = df['case_id'].values
print(doctors)
print(case_id)
```

```python
# get 20% test data from each doctor
import pandas as pd
import numpy as np
split = 0.2
print('total_test:',len(case_id)*split)
test = []
for doctor in doctors:
    cases = df[df['doctor'] == doctor]['case_id'].values
    num_test = int(len(cases) * split)
    if num_test == 0:
        num_test = 1
    num_train = len(cases) - num_test
    
    test_cases = cases[:num_test]
    other_cases = cases[num_test:]
    test.append(test_cases)
test = np.concatenate(test)
print('real test:',test)

# save test data in doctor_category.csv
df['test'] = df['case_id'].apply(lambda x: 1 if x in test else 0)
print(df.head(10))
df.to_csv('../data/OSIC/doctor_category.csv', index=False)
```

# Get test imgs name for OISC dataset

```python
import os
import pandas as pd
# Load test set case IDs from CSV
df = pd.read_csv('/media/NAS06/gavinyue/genai-wsss/data/OSIC/doctor_category.csv')
caseids_test = df['case_id'][df['test'] == 1].tolist()
caseids_test = [str(caseid).zfill(3) for caseid in caseids_test]

img_dir = '/media/NAS06/gavinyue/genai-wsss/data/OSIC/processed/fibrosis'
split_save_dir = '/media/NAS06/gavinyue/genai-wsss/data/OSIC/split/splits_orig_name'
for caseid in caseids_test:
    file_names = sorted(os.listdir(img_dir))
    file_names = [name for name in file_names if name.startswith(caseid) and name.endswith(('.png', '.jpg', '.jpeg'))]
    
    save_path = os.path.join(split_save_dir, 'test.txt')
    # add file_names in text file
    with open(save_path, 'a') as f:
        for file_name in file_names:
            f.write(f'{file_name}\n')
```
