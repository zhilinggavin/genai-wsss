## Visualise 3D scans data

```python
import os
import matplotlib.pyplot as plt
import nibabel as nib
# load_path = '../data/OSIC/raw/64.nii.gz'
load_path = '../data/AIPFR/3d_visual/02_00019.nii.gz'
# load_path = '../data/OSIC/raw/129.nii.gz'

ct_scan = nib.load(load_path)
data = ct_scan.get_fdata()
print(data.shape)
print(f"data min: {data.min()}, max: {data.max()}")
```

```python
depth_id = [i for i, dim in enumerate(data.shape) if data.shape.count(dim) == 1][0]

slice_id = data.shape[depth_id] // 2

# Dynamically create slicing for the correct depth_id
slices = [slice(None)] * len(data.shape)  # Create a list of slices for all dimensions
slices[depth_id] = slice_id  # Set the slice_id in the correct dimension
plt.imshow(data[tuple(slices)], cmap='gray')
```
