```python
import os, glob
import pandas as pd
doctor_dir = '/media/NAS04/yyfang/prognostic_result/dataset/data_fibrosis/annotation_all/fibrosis_Yakup'
save_dir = '/media/NAS06/gavinyue/genai-wsss/data/OSIC'
```

```python
files = glob.glob(doctor_dir + '/*fibrosis.nii.gz')
files = [os.path.basename(f) for f in files]
print('total patient cases:',len(files))
print(files[0])
# get patient case_id
cases_id = [f.split('_fib')[0] for f in files]
print(cases_id[:5])
```

```python
# create dataframe
df = pd.read_csv(os.path.join(save_dir, 'doctor_category.csv'))

new_df = pd.DataFrame({'case_id': cases_id, 'doctor': doctor_dir.split('_')[-1]})
df = pd.concat([df, new_df], axis=0)

df.to_csv(os.path.join(save_dir, 'doctor_category.csv'), index=False)
```

```python
# reorder raws
df = pd.read_csv(os.path.join(save_dir, 'doctor_category.csv'))
df = df.sort_values(by='case_id')
print(df.head())
df.to_csv(os.path.join(save_dir, 'doctor_category.csv'), index=False)
```
